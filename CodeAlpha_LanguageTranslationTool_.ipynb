```python
import ipywidgets as widgets
from IPython.display import display, clear_output
from deep_translator import GoogleTranslator

# Title
title = widgets.HTML(
    "<h2 style='color:#173b70'>🌍 Language Translation Tool</h2>"
)

# Input text
input_text = widgets.Textarea(
    placeholder="Enter the text you want to translate...",
    description="Text:",
    layout=widgets.Layout(width="90%", height="120px")
)

# Language options
languages = {
    "English": "en",
    "Tamil": "ta",
    "Hindi": "hi",
    "French": "fr",
    "Spanish": "es",
    "German": "de",
    "Japanese": "ja",
    "Telugu": "te",
    "Malayalam": "ml"
}

source_language = widgets.Dropdown(
    options=languages,
    value="en",
    description="From:"
)

target_language = widgets.Dropdown(
    options=languages,
    value="ta",
    description="To:"
)

translate_button = widgets.Button(
    description="Translate",
    button_style="primary",
    icon="language"
)

output_text = widgets.Textarea(
    placeholder="Your translated text will appear here...",
    description="Result:",
    layout=widgets.Layout(width="90%", height="120px")
)

status = widgets.HTML()

def translate_text(button):
    try:
        if not input_text.value.strip():
            status.value = "<p style='color:red'>Please enter some text.</p>"
            return

        if source_language.value == target_language.value:
            status.value = "<p style='color:red'>Choose two different languages.</p>"
            return

        status.value = "<p>Translating...</p>"

        result = GoogleTranslator(
            source=source_language.value,
            target=target_language.value
        ).translate(input_text.value)

        output_text.value = result or ""
        status.value = "<p style='color:green'>Translation completed!</p>"

    except Exception as error:
        status.value = (
            "<p style='color:red'>Translation failed. "
            "Check your internet connection and try again.</p>"
        )
        print("Error details:", error)

translate_button.on_click(translate_text)

display(
    title,
    input_text,
    widgets.HBox([source_language, target_language]),
    translate_button,
    output_text,
    status
)
```

```python

import requests
import ipywidgets as widgets
from IPython.display import display, clear_output

# Application title
title = widgets.HTML(
    value="""
    <h1 style="color:#173b70; text-align:center;">
        Language Translation Tool
    </h1>
    <p style="text-align:center;">
        Translate text between multiple languages
    </p>
    """
)

# Supported languages
languages = {
    "English": "en",
    "Tamil": "ta",
    "Hindi": "hi",
    "Telugu": "te",
    "Malayalam": "ml",
    "Kannada": "kn",
    "French": "fr",
    "Spanish": "es",
    "German": "de",
    "Japanese": "ja"
}

# Source and target language dropdowns
source_language = widgets.Dropdown(
    options=list(languages.keys()),
    value="English",
    description="From:",
    layout=widgets.Layout(width="48%")
)

target_language = widgets.Dropdown(
    options=list(languages.keys()),
    value="Tamil",
    description="To:",
    layout=widgets.Layout(width="48%")
)

language_row = widgets.HBox(
    [source_language, target_language],
    layout=widgets.Layout(justify_content="space-between")
)

# Input text area
input_text = widgets.Textarea(
    value="Hello, how are you?",
    placeholder="Enter text to translate...",
    description="Input:",
    layout=widgets.Layout(width="100%", height="120px")
)

# Translate button
translate_button = widgets.Button(
    description="Translate",
    button_style="primary",
    icon="language",
    layout=widgets.Layout(width="160px", height="40px")
)

# Clear button
clear_button = widgets.Button(
    description="Clear",
    button_style="warning",
    layout=widgets.Layout(width="100px", height="40px")
)

# Output text area
output_text = widgets.Textarea(
    value="",
    placeholder="Your translated text will appear here...",
    description="Output:",
    disabled=True,
    layout=widgets.Layout(width="100%", height="120px")
)

# Status message
status = widgets.HTML(value="Ready to translate.")

# Translation function
def translate_text(button=None):
    text = input_text.value.strip()

    if not text:
        output_text.value = ""
        status.value = (
            "<p style='color:red;'>Please enter text to translate.</p>"
        )
        return

    source_code = languages[source_language.value]
    target_code = languages[target_language.value]

    if source_code == target_code:
        output_text.value = text
        status.value = (
            "<p style='color:green;'>Source and target languages are the same.</p>"
        )
        return

    translate_button.disabled = True
    output_text.value = ""
    status.value = "<p style='color:#173b70;'>Translating... Please wait.</p>"

    try:
        response = requests.get(
            "https://api.mymemory.translated.net/get",
            params={
                "q": text,
                "langpair": f"{source_code}|{target_code}"
            },
            timeout=20
        )

        response.raise_for_status()
        data = response.json()

        if data.get("responseStatus") == 200:
            translated = data.get("responseData", {}).get(
                "translatedText", ""
            )

            if translated:
                output_text.value = translated
                status.value = (
                    "<p style='color:green;'>Translation successful!</p>"
                )
            else:
                status.value = (
                    "<p style='color:red;'>No translation returned.</p>"
                )
        else:
            message = data.get(
                "responseDetails", "Translation service unavailable."
            )
            status.value = (
                f"<p style='color:red;'>Translation failed: {message}</p>"
            )

    except requests.exceptions.Timeout:
        status.value = (
            "<p style='color:red;'>Request timed out. Please try again.</p>"
        )

    except requests.exceptions.RequestException as error:
        status.value = (
            f"<p style='color:red;'>Connection error: {error}</p>"
        )

    except Exception as error:
        status.value = (
            f"<p style='color:red;'>Unexpected error: {error}</p>"
        )

    finally:
        translate_button.disabled = False

# Clear function
def clear_fields(button=None):
    input_text.value = ""
    output_text.value = ""
    status.value = "<p>Ready to translate.</p>"

# Connect buttons to functions
translate_button.on_click(translate_text)
clear_button.on_click(clear_fields)

# Display application
app = widgets.VBox(
    [
        title,
        language_row,
        input_text,
        widgets.HBox(
            [translate_button, clear_button],
            layout=widgets.Layout(justify_content="center")
        ),
        output_text,
        status
    ],
    layout=widgets.Layout(
        width="100%",
        max_width="850px",
        margin="auto",
        padding="20px",
        gap="15px"
    )
)

display(app)
```
